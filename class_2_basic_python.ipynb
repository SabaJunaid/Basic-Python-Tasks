## What is Python?
Python is a high-level, interpreted, and easy-to-learn programming language.
It was created by** Guido van Rossum **and released in 1991.

Python is famous because its syntax is simple and easy to understand, which makes it an excellent language for beginners.

## What is Syntax?
Syntax means the set of rules used to write programs in a programming language.

Python syntax is simple and readable.

## logical operatorts

```python
# and= both condition true
# or= any one condition must be true
# not true=false
```

```python

x = 5

# and (Returns True if BOTH conditions are true)
print(x > 3 and x < 10)  # True and True -> Output: True

# or (Returns True if AT LEAST ONE condition is true)
print(x > 10 or x < 4)   # False or False -> Output: False
print(x > 3 or x < 2)    # True or False -> Output: True

# not (Reverses the Boolean result)
print(not(x > 3))        # not(True) -> Output: False
```
